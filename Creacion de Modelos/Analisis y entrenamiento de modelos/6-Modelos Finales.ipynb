## Modelo General

```python
def modelo_fusionado(datos: pl.DataFrame):
      lista_modelos = []
      for cluster in range(1, 18):
          modelo= XGBRegressor()
          modelo.load_model(fname= f"{ruta_base}/Modelos Finales/cluster_{cluster}.json")
          lista_modelos.append(modelo)

      # Extraemos la columna de clústeres y variables predictoras
      clusters = datos.select(["cluster"]).to_numpy()
      caracteristicas = datos.drop(["cluster"])

      # Creamos un array vacio de predicciones para ir rellenando conforme se hagan las predicciones (Asi manetenmos el orden)
      predicciones = np.full(shape= datos.shape[0], fill_value= np.nan)

      # Realizamos las predicciones con base en cada cluster
      for num_modelo, modelo in enumerate(lista_modelos):
          # Creamos una máscara para rellenar con ella el array de nulos y para seleccionar los datos que necesitamos
          mascara_cluster = (clusters == (num_modelo + 1)) # Sumamos 1 porque enumerate los enumera desde 0
          mascara_cluster = mascara_cluster.flatten() # Aplanamos la máscara bidimensional

          # Sí hay filas con respecto a este cluster, predecimos únicamente estas filas
          if mascara_cluster.any():
              predicciones[mascara_cluster] = modelo.predict(caracteristicas.filter(mascara_cluster).to_pandas())

      return predicciones
```

```python
datos = df.filter(pl.col("año") == 2017).drop(["unidades_vendidas", "año"]).collect(engine= "streaming")
predicciones = modelo_fusionado(datos = datos)
```

## Métricas Modelo General

```python
#Valores Reales
valores_reales = df.select(["año", "unidades_vendidas", "perecedero"]).filter(pl.col("año") == 2017).drop(["año"]).collect(engine="streaming").to_pandas()

# Para ver metricas generales del modelo general
error_absoluto_medio            = mean_absolute_error(y_pred=predicciones, y_true=valores_reales["unidades_vendidas"])
raiz_error_cuadratico_medio     = root_mean_squared_error(y_pred=predicciones, y_true=valores_reales["unidades_vendidas"])
r2                              = r2_score(y_pred=predicciones, y_true=valores_reales["unidades_vendidas"])
error_maximo                    = max_error(y_pred=predicciones, y_true=valores_reales["unidades_vendidas"])
mediana_error_absoluto          = median_absolute_error(y_pred=predicciones, y_true=valores_reales["unidades_vendidas"])
nwrmsle_metrica                 = nwrmsle(y_true=valores_reales["unidades_vendidas"],y_pred=predicciones,perecedero=valores_reales["perecedero"])

diccionario_metricas = {
    "Mean Absolute Error": error_absoluto_medio,
    "RMSE" : raiz_error_cuadratico_medio,
    "Median Absolute Error" : mediana_error_absoluto,
    "Max Error" : error_maximo,
    "R2": r2,
    "NWRMSLE": nwrmsle_metrica
}

# Para ver metricas por bloques de unidades Vendidas del modelo general
bloques = [
             ("0", 0, 0),
            ("1-5", 1, 5),
            ("6-20", 6, 20),
            ("21-50", 21, 50),
            ("51-100", 51, 100),
            ("101-500", 101, 500),
            (">500", 501, np.inf)
        ]
y_real = valores_reales["unidades_vendidas"]
y_real_perecedero = valores_reales["perecedero"]
resultados_bloques = []
for nombre_bloque, minimo, maximo in bloques:
    mascara     = ((y_real >= minimo) &(y_real <= maximo))
    y_bloque    = y_real[mascara]
    perecedero_bloque = y_real_perecedero[mascara]
    pred_bloque = predicciones[mascara]

    if len(y_bloque) == 0:
        continue

    resultados_bloques.append({
                "bloque"          : nombre_bloque,
                 "Ventas Reales"   : y_bloque.sum(),
                "Ventas Predichas": pred_bloque.sum(),
                "MAE"             : mean_absolute_error(y_bloque, pred_bloque),
                "NWRMSLE"         : nwrmsle(y_true=y_bloque, y_pred=pred_bloque, perecedero=perecedero_bloque.to_numpy())
            })
```

### Conclusiones Rendimiento General del Modelo

El rendimiento del modelo es aceptable en algunos puntos, pero flaquea en otros. Predecir las unidades vendidas de un producto en una cadena de supermercado resulta complicado por la cantidad de factores externos que influyen en el comportamiento de compras de los consumidores.

- Se presenta un MAE de 2.14, lo que considerando la distribucion de los datos parece malo, pero esta es una métrica que se ve muy afectada por valores atos debido a su naturaleza de media, lo que implica para su interpretacion correcta apoyarse de otra metrica basada en la mediana que es la mediana del error absoluto, donde podemos observar que aproximadamente más de la mitad de los datos recibieron un error menor a 1 (0.66), lo cual si tiene más concordancia con la distribucion de los datos. Esto indica que el modelo tiende a estar muy cerca de su valor verdadero.

- Se presenta un RMSE de 13.82, lo que indica que el modelo puede tener una MAE un poco confiable, pero llega a cometer errores muy grandes. Analizando la distribucion de los datos para cada cluster independiente, podemos observar que todos los modelos tuvieron problemas para que su prediccion maxima se acerque a las unidades vendidas más altas de su respectivo cluster. Esto se debe a que el modelo aprende mejor a predecir valores bajos por el exceso de ceros.

- Se presenta un R2 de 0.34, lo que indica que una poca variabilidad en las predicciones por las variables predictoras. Aunque esta metrica no está directamente relacionada con el rendimiento del modelo.

- Se presenta un NWRMSLE de 0.55, lo que púede ser aceptable. Esto nos quiere decir que nuestro modelo se equivoca un 55% en la prediccion de unidades vendidas, cuidando más el inventario de los productos perecibles

### Conclusiones Rendimiento General del Modelo Analizando por Bloques de Unidades Vendidas

#### 0 Unidades Vendidas
- El modelo resulta tener un rendimiento aceptable en su MAE y NWRMSLE cuando se trata de predecir productos con cero unidades vendidas. Aunque parece predecir muchas unidades vendidas con respecto a las reales esto principalmente sucede porque al cometer varios errores pequeños (como 0.0002), estos se van sumando y aumentan la cantidad con respecto a la real. esto puede solucionarse con un modelo de dos estapas que detecte si un producto se va a vender o no, funcionando como filtro y asi evitar esos pequeños errores. Además, esto hace que el modelo sea util par analisis, más no para producción.

#### 1-5 Unidades Vendidas
- El modelo resulta tener su mejor rendimiento cuando de predecir unidades bajas se trata, presentando un MAE de 1.5 y un NWRMSLE de 0.51, no obstante, suele predecir un poco más de lo esperado, lo que puede producir un exceso de stock, pero también hay que tener en cuenta que su NWRMSLE nos indica que gran parte de esos productos no serían perecederos

#### 6-20 Unidades Vendidas
- El modelo comienza a flaquear su rendimiento desde este punto, donde presenta un resultado aceptable en su MAE con respecto a las unidades vendidas, pero predice menos de lo esperado en las unidades vendidas, lo que podría producir substock de diversos productos

#### 21-50 y 51-100 Unidades Vendidas
- El modelo ya flaquea bastante con un NWRMSLE muy alto, pero aún presenta un rendimiento aceptable en su MAE lo que lo hace util para predecir estas unidades vendidas.

#### > 100 Unidades Vendidas
- Predecir valores elevados es la mayor debilidad del modelo. Esto surge por la gran presencia de 0 (por los productos que no se venden) y por la poca presencia de productos elevados, lo que hace que el modelo prefiera ser un poco más conservador y  no querer arriesgarse a predecir valores altos
