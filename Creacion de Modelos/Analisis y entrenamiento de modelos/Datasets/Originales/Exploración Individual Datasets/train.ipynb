```python
import pandas as pd
from pathlib import Path
ruta_base = Path.cwd().parent
ruta_base
```

```python
df = pd.read_csv(filepath_or_buffer= ruta_base / "train.csv", parse_dates= ["date"], low_memory= False, dtype_backend= "pyarrow")
df.head()
```

```python
print("====== Columna 'date' ======")
print()
print("minimo: ", df["date"].min())
print("maximo: ", df["date"].max())
```

```python
print("====== Valores Nulos ======")
print()
print(f"id          : {df['id'].isnull().sum()}")
print(f"date        : {df['date'].isnull().sum()}")
print(f"store_nbr   : {df['store_nbr'].isnull().sum()}")
print(f"item_nbr    : {df['item_nbr'].isnull().sum()}")
print(f"unit_sales  : {df['unit_sales'].isnull().sum()}")
print(f"onpromotion : {df['onpromotion'].isnull().sum()}")
```

```python
print("====== Valores Unicos ======")
print()
print(f"store_nbr   : {df['store_nbr'].unique()}")
print(f"item_nbr    : {df['item_nbr'].unique()}")
print(f"onpromotion : {df['onpromotion'].unique()}")
```
