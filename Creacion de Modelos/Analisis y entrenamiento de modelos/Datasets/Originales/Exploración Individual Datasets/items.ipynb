```python
import pandas as pd
import matplotlib.pyplot as plt
from FuncionResumenDataset import resumen_dataset
from pathlib import Path
ruta_base = Path.cwd().parent
ruta_base
```

```python
df = pd.read_csv(filepath_or_buffer= ruta_base/ "items.csv")
df.head()
```

```python
resumen_dataset(dataframe= df, columnas_excluir=["item_nbr"])
```

```python
print("Cantidad de productos distintos: ", df["item_nbr"].nunique())
print("Cantidad de clases distintas   : ", df["class"].nunique())
```

```python
plt.figure(figsize=(12, 8))
plt.title("Distribución 'family'")
grafico = plt.bar(df["family"].value_counts().index ,df["family"].value_counts())
plt.bar_label(grafico)
plt.xlabel("Familia del producto")
plt.ylabel("Cantidad de productos diferentes")
plt.xticks(rotation= 90)
plt.tight_layout()
plt.show()
```

```python
plt.title("Distribución 'perishable'")
grafico= plt.bar(list(map(str, df["perishable"].value_counts().index)), df["perishable"].value_counts())
plt.bar_label(grafico)
plt.xlabel("Etiqueta de perecedero")
plt.ylabel("Cantidad de productos")
plt.tight_layout()
plt.show()
```
