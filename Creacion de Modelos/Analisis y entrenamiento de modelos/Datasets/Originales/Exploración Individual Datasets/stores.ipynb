```python
import pandas as pd
from FuncionResumenDataset import resumen_dataset
import matplotlib.pyplot as plt
from pathlib import Path
ruta_base = Path.cwd().parent
ruta_base
```

```python
df = pd.read_csv(filepath_or_buffer= ruta_base/"stores.csv")
df.head()
```

```python
resumen_dataset(dataframe= df, columnas_excluir=[])
```

```python
plt.title("Cantidad de tiendas por su tipo")
grafico= plt.bar(df["type"].value_counts().index, df["type"].value_counts())
plt.bar_label(grafico)
plt.xlabel("Tipo")
plt.ylabel("Número de tiendas")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))
plt.title("Cantidad de tiendas por su estado")
grafico = plt.bar(df["state"].value_counts().index, df["state"].value_counts())
plt.bar_label(grafico)
plt.xlabel("Estado")
plt.ylabel("Número de tiendas")
plt.xticks(rotation= 90)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))
plt.title("Cantidad de tiendas por ciudad")
grafico = plt.bar(df["city"].value_counts().index, df["city"].value_counts())
plt.bar_label(grafico)
plt.xlabel("Ciudad")
plt.ylabel("Número de tiendas")
plt.xticks(rotation= 90)
plt.tight_layout()
plt.show()
```
