```python
import pandas as pd
import matplotlib.pyplot as plt
from FuncionResumenDataset import resumen_dataset
from pathlib import Path
ruta_base = Path.cwd().parent
ruta_base
```

```python
df = pd.read_csv(filepath_or_buffer= ruta_base/"holidays_events.csv", parse_dates= ["date"])
df.head()
```

```python
resumen_dataset(dataframe= df, columnas_excluir= ["date"])
```

```python
print("====== Columna 'date' ======")
print("minimo: ", df["date"].min())
print("maximo: ", df["date"].max())
```

```python
plt.title("Distribución tipos de feriados en Ecuador")
grafico = plt.bar(df["locale"].value_counts().index, df["locale"].value_counts())
plt.bar_label(grafico)
plt.xlabel("Tipo feriado")
plt.ylabel("Cantidad de feriados")
plt.tight_layout()
plt.show()
```

```python
df_nacional = df[df["locale"]== "National"]
df_local = df[df["locale"]== "Local"]
df_regional = df[df["locale"]== "Regional"]

fig, axs = plt.subplots(ncols=3, figsize= (17,5))
axs[0].set_title("Distribución de localidad según feriados National")
grafico1 = axs[0].bar(df_nacional["locale_name"].value_counts().index, df_nacional["locale_name"].value_counts())
axs[0].bar_label(grafico1)
axs[0].set_xlabel("Localidades")
axs[0].set_ylabel("Numero de Feriados")

axs[1].set_title("Distribución de localidad según feriados Local")
grafico2 = axs[1].bar(df_local["locale_name"].value_counts().index, df_local["locale_name"].value_counts())
axs[1].bar_label(grafico2)
axs[1].set_xlabel("Localidades")
axs[1].set_ylabel("Numero de Feriados")
axs[1].tick_params(axis='x', labelrotation=90)


axs[2].set_title("Distribución de localidad según feriados Regional")
grafico3 = axs[2].bar(df_regional["locale_name"].value_counts().index, df_regional["locale_name"].value_counts())
axs[2].bar_label(grafico3)
axs[2].set_xlabel("Localidades")
axs[2].set_ylabel("Numero de Feriados")
axs[2].tick_params(axis='x', labelrotation=90)
```
