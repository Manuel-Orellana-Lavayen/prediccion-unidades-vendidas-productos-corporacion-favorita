```python
import pandas as pd
from pathlib import Path
ruta_base = Path.cwd().parent.parent
ruta_base
```

```python
df = pd.read_csv(filepath_or_buffer= ruta_base/"Analisis y entrenamiento de modelos/Datasets/Modificados/oil_limpio.csv", parse_dates=["date"])
df = df[(df["date"] >= "2016-09-01") & (df["date"] <= "2017-08-31") ]
df = df.sort_values(by="date")
df = df.rename(columns={"date": "fecha",
                        "dcoilwtico": "precio_petroleo"})
df.head()
```

```python
df["petroleo_promedio_3meses"] = (
    df["precio_petroleo"].shift(1)
    .rolling(window= 90, min_periods= 1)
    .mean()
    .fillna(df["precio_petroleo"])
    .astype("float32")
)
df = df.drop(columns=["precio_petroleo"])
```

```python
df.to_csv(ruta_base/"CreacionBaseConocimiento/BaseConocimiento/petroleo_promedio_3meses.csv", index= None)
```
